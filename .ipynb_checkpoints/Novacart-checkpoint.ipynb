# ***<p style="text-align:center; font-size:250%; font-family:'Times New Roman'; color:green;"> Data Overview:</p>***



# ***The NovaCart dataset contains e-commerce order, delivery, product, customer, return, refund, replacement, and return-cost information for the year 2025. The dataset is divided into multiple interconnected tables, which helps in performing detailed analysis from different business perspectives.***

# ***The main tables used in this analysis are **Customers, Products, Warehouses, Orders, Delivery, Returns, Missing Items, Refunds, and Return Costs**. These tables are connected using common identifiers such as Customer_ID, Product_ID, Warehouse_ID, and Order_ID.***

# ***The **Customers** table contains customer information such as customer segment, registration date, zone, and area. The **Products** table provides product details including category, sub-category, price, rating, and seller information. The **Warehouses** table contains warehouse details along with their zones and areas.***

# ***The **Orders** table contains order-level information such as order date, quantity, discount, order value, payment method, and order status. The **Delivery** table contains courier information, shipping method, expected delivery date, actual delivery date, and delivery status, which helps analyze delivery delays and their relationship with product returns.***

# ***The **Returns** table contains return status, return date, return reason, return type, product condition, and inspection status. The **Missing Items** table helps identify orders where items were missing and understand their contribution to returns.***

# ***The **Refunds** table contains refund status, refund date, refund amount, and refund method. The **Return Costs** table provides information about return shipping costs, replacement costs, processing costs, restocking costs, and recoverable amounts.***

# ***This interconnected dataset allows NovaCart to perform Exploratory Data Analysis (EDA) on return patterns, delivery performance, product and category performance, warehouse and location-wise returns, refund and replacement activity, and the overall financial impact of product returns.***


# ***<p style="text-align:center; font-size:250%; font-family:'Times New Roman'; color:green;">Business Problem </p>***



# ***NovaCart is facing a high volume of product returns and delivery delays, which are increasing operational costs and affecting overall profitability. The business does not have clear visibility into the main reasons behind returns or whether poor delivery performance is contributing to them.***

# ***The company needs to identify the major factors responsible for product returns, including delivery delays, product categories, return reasons, warehouses, locations, and courier performance. It also needs to understand the financial impact of refunds, replacements, and return-related costs.***

# ***The objective is to identify the key drivers of returns and delivery-related issues so that NovaCart can reduce unnecessary returns, improve delivery performance, control operational costs, and improve overall profitability.***

# ***Create a function for all table***

```python
def basic_eda(data, name):
    print("=" * 50)
    print(f"EDA REPORT: {name}")
    print("=" * 50)
    
    print("\nShape:")
    print(data.shape)
    
    print("\nFirst 5 Rows:")
    display(data.head())
    
    print("\nData Types:")
    print(data.dtypes)
    
    print("\nMissing Values:")
    print(data.isnull().sum())
    
    print("\nDuplicate Rows:")
    print(data.duplicated().sum())
    
    print("\nNumerical Summary:")
    display(data.describe())
    
    print("\nCategorical Summary:")
    display(data.describe(include='object'))
```

# ***Convert Object-Type Date Columns to Datetime Format***

```python
def convert_dates(data, date_columns):
    for col in date_columns:
        data[col] = pd.to_datetime(data[col], errors='coerce')
    return data
```

```python
customers = convert_dates(
    customers,
    ['Customer_Registration_Date']
)
orders = convert_dates(
    orders,
    ['Order_Date']
)
delivery = convert_dates(
    delivery,
    ['Expected_Date', 'Actual_Date']
)
returns = convert_dates(
    returns,
    ['Return_Date']
)
refunds = convert_dates(
    refunds,
    ['Refund_Date']
)
```

## ***Return Reason Distribution***

```python

return_reason = (
    returns['Return_Reason']
    .value_counts()
    .reset_index()
)

return_reason.columns = ['Return_Reason', 'Total_Returns']

return_reason['Return_Percentage'] = (
    return_reason['Total_Returns']
    / return_reason['Total_Returns'].sum()
    * 100
)

return_reason['Return_Percentage'] = (
    return_reason['Return_Percentage'].round(2)
)
```

```python

plt.figure(figsize=(10, 6))

plt.bar(
    return_reason['Return_Reason'],
    return_reason['Total_Returns']
)

plt.title('Major Reasons for Product Returns')
plt.xlabel('Return Reason')
plt.ylabel('Number of Returns')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    return_reason['Total_Returns'],
    labels=return_reason['Return_Reason'],
    autopct='%1.1f%%'
)

plt.title('Percentage Distribution of Product Return Reasons')
plt.show()
```

# ***Return Reason by Product Category***

```python
# Merge Returns with Orders

returns_orders = returns.merge(
    orders[['Order_ID', 'Product_ID']],
    on='Order_ID',
    how='left'
)

# Merge with Products

returns_products = returns_orders.merge(
    products[['Product_ID', 'Category', 'Sub_Category']],
    on='Product_ID',
    how='left'
)

returns_products.head()
```

```python
# Return Reasons by Product Category

category_reason = pd.crosstab(
    returns_products['Category'],
    returns_products['Return_Reason']
)

category_reason
```

```python
# Calculate Return Reason Percentage within Each Category

category_reason_pct = (
    category_reason
    .div(category_reason.sum(axis=1), axis=0)
    * 100
)

category_reason_pct = category_reason_pct.round(2)

category_reason_pct
```

```python

plt.figure(figsize=(12, 7))

sns.heatmap(
    category_reason_pct,
    annot=True,
    fmt='.1f'
)

plt.title('Return Reason Percentage by Product Category')
plt.xlabel('Return Reason')
plt.ylabel('Product Category')
plt.tight_layout()
plt.show()
```


# ***<p style="font-family:times new  roman;color:green;font-size: 250%; text-align:center">Objective 2 — Analyze the Relationship Between Delivery Delays and Product Returns</p>***


# ***Calculate Delivery Delay***

```python


delivery['Delivery_Delay_Days'] = (
    delivery['Actual_Date'] - delivery['Expected_Date']
).dt.days

delivery[['Order_ID', 'Expected_Date', 'Actual_Date', 'Delivery_Delay_Days']].head(10)
```

```python


delivery['Delivery_Performance'] = delivery['Delivery_Delay_Days'].apply(
    lambda x: 'Delayed' if x > 0 else 'On Time'
)

delivery[
    ['Order_ID', 'Delivery_Delay_Days', 'Delivery_Performance']
].head()
```

```python


delivery_performance = (
    delivery['Delivery_Performance']
    .value_counts()
    .reset_index()
)

delivery_performance.columns = [
    'Delivery_Performance',
    'Total_Orders'
]

delivery_performance['Percentage'] = (
    delivery_performance['Total_Orders']
    / delivery_performance['Total_Orders'].sum()
    * 100
).round(2)

delivery_performance
```

```python


plt.figure(figsize=(8, 5))

plt.bar(
    delivery_performance['Delivery_Performance'],
    delivery_performance['Total_Orders']
)

plt.title('Overall Delivery Performance')
plt.xlabel('Delivery Performance')
plt.ylabel('Number of Orders')

plt.tight_layout()
plt.show()
```

```python
# Distribution of Delivery Delays

delayed_orders = delivery[
    delivery['Delivery_Delay_Days'] > 0
]

delayed_orders['Delivery_Delay_Days'].describe()
```

```python
# Delivery Delay Distribution

plt.figure(figsize=(10, 6))

plt.hist(
    delayed_orders['Delivery_Delay_Days'],
    bins=20
)

plt.title('Distribution of Delivery Delays')
plt.xlabel('Delay in Days')
plt.ylabel('Number of Orders')

plt.tight_layout()
plt.show()
```

```python


def classify_delay(days):
    if days <= 0:
        return 'On Time'
    elif days <= 2:
        return '1-2 Days'
    elif days <= 5:
        return '3-5 Days'
    elif days <= 7:
        return '6-7 Days'
    else:
        return 'More than 7 Days'


delivery['Delay_Category'] = delivery[
    'Delivery_Delay_Days'
].apply(classify_delay)

delivery['Delay_Category'].value_counts()
```

```python


delay_category = (
    delivery['Delay_Category']
    .value_counts()
    .reindex([
        'On Time',
        '1-2 Days',
        '3-5 Days',
        '6-7 Days',
        'More than 7 Days'
    ])
)

plt.figure(figsize=(10, 6))

plt.bar(
    delay_category.index,
    delay_category.values
)

plt.title('Delivery Delay Categories')
plt.xlabel('Delay Category')
plt.ylabel('Number of Orders')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

# ***Merge Delivery with Return Data***

```python
# Create Return Flag

return_flag = returns[['Order_ID']].drop_duplicates()

return_flag['Returned'] = 'Yes'

# Merge Delivery with Return Information

delivery_return = delivery.merge(
    return_flag,
    on='Order_ID',
    how='left'
)

delivery_return['Returned'] = delivery_return['Returned'].fillna('No')

delivery_return[
    ['Order_ID', 'Delivery_Performance', 'Delay_Category', 'Returned']
].head()
```

```python
# Calculate Return Rate by Delivery Performance

delivery_return_rate = (
    delivery_return
    .groupby('Delivery_Performance')
    .agg(
        Total_Orders=('Order_ID', 'count'),
        Returned_Orders=('Returned', lambda x: (x == 'Yes').sum())
    )
    .reset_index()
)

delivery_return_rate['Return_Rate'] = (
    delivery_return_rate['Returned_Orders']
    / delivery_return_rate['Total_Orders']
    * 100
).round(2)

delivery_return_rate
```

```python
# Return Rate: On-Time vs Delayed

plt.figure(figsize=(8, 5))

plt.bar(
    delivery_return_rate['Delivery_Performance'],
    delivery_return_rate['Return_Rate']
)

plt.title('Return Rate by Delivery Performance')
plt.xlabel('Delivery Performance')
plt.ylabel('Return Rate (%)')

plt.tight_layout()
plt.show()
```

# ***Courier-wise Delivery and Return Analysis***

```python
# Courier-wise Delivery and Return Analysis

courier_analysis = (
    delivery_return
    .groupby('Courier')
    .agg(
        Total_Orders=('Order_ID', 'count'),
        Delayed_Orders=('Delivery_Performance', lambda x: (x == 'Delayed').sum()),
        Returned_Orders=('Returned', lambda x: (x == 'Yes').sum())
    )
    .reset_index()
)

courier_analysis['Delay_Rate'] = (
    courier_analysis['Delayed_Orders']
    / courier_analysis['Total_Orders']
    * 100
).round(2)

courier_analysis['Return_Rate'] = (
    courier_analysis['Returned_Orders']
    / courier_analysis['Total_Orders']
    * 100
).round(2)

courier_analysis.sort_values(
    'Delay_Rate',
    ascending=False
)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    courier_analysis['Courier'],
    courier_analysis['Delay_Rate']
)

plt.title('Courier-wise Delivery Delay Rate')
plt.xlabel('Courier')
plt.ylabel('Delay Rate (%)')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    courier_analysis['Courier'],
    courier_analysis['Return_Rate']
)

plt.title('Courier-wise Return Rate')
plt.xlabel('Courier')
plt.ylabel('Return Rate (%)')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

# ***Shipping Method vs Delivery & Returns***

```python


shipping_analysis = (
    delivery_return
    .groupby('Shipping_Method')
    .agg(
        Total_Orders=('Order_ID', 'count'),
        Delayed_Orders=('Delivery_Performance', lambda x: (x == 'Delayed').sum()),
        Returned_Orders=('Returned', lambda x: (x == 'Yes').sum())
    )
    .reset_index()
)

shipping_analysis['Delay_Rate'] = (
    shipping_analysis['Delayed_Orders']
    / shipping_analysis['Total_Orders']
    * 100
).round(2)

shipping_analysis['Return_Rate'] = (
    shipping_analysis['Returned_Orders']
    / shipping_analysis['Total_Orders']
    * 100
).round(2)

shipping_analysis
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    shipping_analysis['Shipping_Method'],
    shipping_analysis['Delay_Rate']
)

plt.title('Delivery Delay Rate by Shipping Method')
plt.xlabel('Shipping Method')
plt.ylabel('Delay Rate (%)')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    shipping_analysis['Shipping_Method'],
    shipping_analysis['Return_Rate']
)

plt.title('Return Rate by Shipping Method')
plt.xlabel('Shipping Method')
plt.ylabel('Return Rate (%)')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

# ***Warehouse-wise Delivery & Return Analysis***

```python


delivery_warehouse = delivery_return.merge(
    orders[['Order_ID', 'Warehouse_ID']],
    on='Order_ID',
    how='left'
)

delivery_warehouse = delivery_warehouse.merge(
    warehouses[['Warehouse_ID', 'Warehouse_Name', 'Zone', 'Area']],
    on='Warehouse_ID',
    how='left'
)

delivery_warehouse.head()
```

```python


warehouse_analysis = (
    delivery_warehouse
    .groupby(['Warehouse_ID', 'Warehouse_Name', 'Zone'])
    .agg(
        Total_Orders=('Order_ID', 'count'),
        Delayed_Orders=('Delivery_Performance', lambda x: (x == 'Delayed').sum()),
        Returned_Orders=('Returned', lambda x: (x == 'Yes').sum())
    )
    .reset_index()
)

warehouse_analysis['Delay_Rate'] = (
    warehouse_analysis['Delayed_Orders']
    / warehouse_analysis['Total_Orders']
    * 100
).round(2)

warehouse_analysis['Return_Rate'] = (
    warehouse_analysis['Returned_Orders']
    / warehouse_analysis['Total_Orders']
    * 100
).round(2)

warehouse_analysis.sort_values(
    'Delay_Rate',
    ascending=False
)
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    warehouse_analysis['Warehouse_Name'],
    warehouse_analysis['Delay_Rate']
)

plt.title('Warehouse-wise Delivery Delay Rate')
plt.xlabel('Warehouse')
plt.ylabel('Delay Rate (%)')

plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    warehouse_analysis['Warehouse_Name'],
    warehouse_analysis['Return_Rate']
)

plt.title('Warehouse-wise Return Rate')
plt.xlabel('Warehouse')
plt.ylabel('Return Rate (%)')

plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

# ***Delivery Delay × Return Reason***

```python
# Merge Delayed Orders with Return Information

delayed_returns = delivery_return[
    delivery_return['Delivery_Performance'] == 'Delayed'
].merge(
    returns[['Order_ID', 'Return_Reason']],
    on='Order_ID',
    how='inner'
)

delayed_returns.head()
```

```python
# Return Reasons for Delayed Orders

delayed_reason = (
    delayed_returns['Return_Reason']
    .value_counts()
    .reset_index()
)

delayed_reason.columns = [
    'Return_Reason',
    'Returned_Orders'
]

delayed_reason['Percentage'] = (
    delayed_reason['Returned_Orders']
    / delayed_reason['Returned_Orders'].sum()
    * 100
).round(2)

delayed_reason
```

```python
# Percentage of Return Reasons by Delivery Performance
reason_delivery = (
    returns
    .merge(
        delivery[['Order_ID', 'Delivery_Status']],
        on='Order_ID',
        how='left'
    )
    .groupby(['Return_Reason', 'Delivery_Status'])
    .size()
    .unstack(fill_value=0)
)


reason_delivery_pct = (
    reason_delivery
    .div(reason_delivery.sum(axis=0), axis=1)
    * 100
).round(2)

reason_delivery_pct
```

```python
plt.figure(figsize=(10, 6))

reason_delivery_pct.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Return Reason Distribution: Delayed vs On-Time Deliveries')
plt.xlabel('Return Reason')
plt.ylabel('Percentage of Returns (%)')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```


# ***<p style="font-family:times new  roman;color:green;font-size: 250%; text-align:center">Objective 5 — Evaluate Courier and Delivery Performance</p>***


# ***Courier Performance***

```python
courier_performance = (
    delivery_return
    .groupby('Courier')
    .agg(
        Total_Orders=('Order_ID', 'count'),
        Delayed_Orders=('Delivery_Performance', lambda x: (x == 'Delayed').sum()),
        Returned_Orders=('Returned', lambda x: (x == 'Yes').sum())
    )
    .reset_index()
)

courier_performance['Delay_Rate'] = (
    courier_performance['Delayed_Orders']
    / courier_performance['Total_Orders'] * 100
).round(2)

courier_performance['Return_Rate'] = (
    courier_performance['Returned_Orders']
    / courier_performance['Total_Orders'] * 100
).round(2)

courier_performance
```

```python
plt.figure(figsize=(10, 6))

x = np.arange(len(courier_performance))
width = 0.35

plt.bar(
    x - width/2,
    courier_performance['Delay_Rate'],
    width,
    label='Delay Rate'
)

plt.bar(
    x + width/2,
    courier_performance['Return_Rate'],
    width,
    label='Return Rate'
)

plt.xticks(x, courier_performance['Courier'])
plt.xlabel('Courier')
plt.ylabel('Rate (%)')
plt.title('Courier Performance: Delay Rate vs Return Rate')
plt.legend()
plt.tight_layout()
plt.show()
```
